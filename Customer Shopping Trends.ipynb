```python
df["Category"].value_counts()
```

### category-wise analysis

Finding:
The analysis shows that Clothing has the highest number of purchases (1,737), followed by Accessories (1,240). Outerwear has the lowest number of purchases (324).

Business Insight:
Customers show more interest in Clothing and Accessories compared to other categories.

Business Decision:
The business can focus on Clothing and Accessories promotions and maintain sufficient stock in these categories. It can also introduce offers to increase Outerwear sales.

```python
df["Category"].value_counts().plot.bar()
plt.title("Category-wise Shopping Count")
plt.xlabel("Category")
plt.ylabel("Number of Purchases")
plt.show()
```

```python
df["Item Purchased"].value_counts().head(10)
```

### Item Purchased

Finding:
The bar chart shows the number of purchases for different items. It helps identify the most and least purchased items.

Business Insight:
The business can understand which individual products are more popular among customers.

Business Decision:
The business can maintain sufficient stock of popular items and introduce promotions for less popular items.

```python
df["Item Purchased"].value_counts().head(10).plot.barh()
plt.title("Top 10 Most Purchased Items")
plt.xlabel("Number of Purchases")
plt.ylabel("Item Purchased")
plt.show()
```

```python
df["Purchase Amount (USD)"].describe()
```

### Purchase Amount Distribution

Finding:
The average purchase amount is approximately $59.76. The minimum purchase amount is $20, and the maximum is $100.

Business Insight:
Customers spend different amounts when shopping, with the average spending close to $60.

Business Decision:
The business can create offers and product bundles around common customer spending levels to encourage more purchases.

```python
plt.hist(df["Purchase Amount (USD)"],bins=15)
plt.title("Purchase Amount Distribution")
plt.xlabel("Purchase Amount (USD)")
plt.ylabel("Number of Customers")
plt.show()
```

```python
df["Gender"].value_counts()
```

### Gender-wise Shopping

Finding:
The chart shows that male customers account for approximately 68% of the records, while female customers account for 32%.

Business Insight:
Male customers represent a larger share of the customers in this dataset.

Business Decision:
The business can study the shopping preferences of both groups and create targeted marketing campaigns.

```python
df["Gender"].value_counts().plot.pie(autopct="%1.1f%%",labels=None)
plt.title("Gender-wise Shopping")
plt.ylabel("")
plt.legend(df["Gender"].value_counts().index)
plt.show()
```

```python
df["Payment Method"].value_counts()
```

### Payment Method Analysis

Finding:
Credit Card is the most frequently used payment method with 696 records, while Bank Transfer has the lowest count with 632 records.

Business Insight:
Customers use different payment methods, but Credit Card is slightly more popular.

Business Decision:
The business should continue supporting multiple payment methods and make the payment process convenient for customers.

```python
df["Payment Method"].value_counts().plot.bar()
plt.title("Payment Wise Analysis")
plt.xlabel("Payment Method ")
plt.ylabel("number of counts")
plt.show()
```

```python
df["Season"].value_counts()
```

### Season-wise Shopping

Finding:
Shopping activity is almost equally distributed across all four seasons. Spring accounts for 25.6%, Fall 25%, Winter 24.9%, and Summer 24.5%.

Business Insight:
Customer shopping activity remains relatively consistent throughout the seasons.

Business Decision:
The business can plan seasonal promotions throughout the year instead of depending on only one season.

```python
SG=df.pivot_table(index="Season",columns="Gender",values="Purchase Amount (USD)",aggfunc="mean")
SG.plot.bar(stacked=True)
plt.title("Season-wise Average Purchase by Gender")
plt.xlabel("Season")
plt.ylabel("Average Purchase Amount")
plt.legend(title="Gender",bbox_to_anchor=(1,1))
plt.show()
```

```python
df.groupby("Category")["Purchase Amount (USD)"].mean().plot.bar()
plt.title("Average Purchase by Category")
plt.xlabel("Category")
plt.ylabel("Average Purchase Amount")
plt.show()
```

```python
df.groupby("Payment Method")["Purchase Amount (USD)"].mean().plot.bar()
plt.title("Average Purchase by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Average Purchase Amount")
plt.show()
```

```python
PC=df.pivot_table(index="Category",columns="Payment Method",values="Purchase Amount (USD)",aggfunc="mean")
PC.plot.bar(stacked=False)
plt.title("Category vs Payment Method")
plt.xlabel("Category")
plt.ylabel("Average Purchase Amount")
plt.legend(bbox_to_anchor=(1,1))
plt.show()
```
